```python
#Create a scatter plot
votes_data = df.dropna(subset=["imdb_score", "imdb_votes"])

plt.figure(figsize=(9, 5))

plt.scatter(
    votes_data["imdb_votes"],
    votes_data["imdb_score"],
    alpha=0.3
)

plt.title("IMDb Scores vs Number of Votes")
plt.xlabel("Number of IMDb Votes")
plt.ylabel("IMDb Score")

plt.tight_layout()
plt.show()
```

observation
Most titles cluster on the left. Most have relatively few votes compared with the handful of titles that have hundreds of thousands or even millions.
- Scores vary widely among titles with fewer votes. In that dense cluster, we can see both low and high ratings.
- Some highly voted titles have high scores. However, the plot alone doesn't prove that more votes lead to higher ratings.
Your chart has one limitation: a small number of titles with extremely high vote counts compresses the majority of points near the left edge.

```python
#Improve the chart using a logarithmic scale
plt.figure(figsize=(9, 5))

plt.scatter(
    votes_data["imdb_votes"],
    votes_data["imdb_score"],
    alpha=0.3
)

plt.xscale("log")
plt.title("IMDb Scores vs Number of Votes (Log Scale)")
plt.xlabel("Number of IMDb Votes (log scale)")
plt.ylabel("IMDb Score")

plt.tight_layout()
plt.show()
#A logarithmic scale will spread out the lower vote counts and make the pattern easier to inspect.
```

Most titles have IMDb scores concentrated roughly between 5 and 8.



The scores vary considerably, particularly among titles with relatively few votes.



At higher vote counts, many titles have scores around 6–8, although some score higher or lower.



There may be a positive relationship between vote count and IMDb score, but the chart alone isn't enough to establish its strength.

```python
#Calculate the correlation
correlation = votes_data["imdb_score"].corr(
    votes_data["imdb_votes"]
)

print("Correlation between IMDb score and votes:", round(correlation, 3))
```

#The correlation between IMDb scores and vote counts was 0.19, indicating a weak positive linear relationship. This suggests that titles with more votes tend to have slightly higher scores, but vote count alone does not explain much of the variation in ratings.

```python
#Separate the genres
# Split multiple genres into individual values
genres = df["genres"].dropna().str.split(", ").explode()

# Count how often each genre appears
genre_counts = genres.value_counts()

# Display the 10 most frequent genres
print(genre_counts.head(10))
#dropna() excludes records without genre information.
#str.split(", ") separates genres wherever a comma and space appear.
#explode() puts each genre into its own row.
#value_counts() counts the occurrences of each genre.
```

```python
#Inspect the original genre values
df["genres"].dropna().head(10).tolist()
```

```python
#extracting individual genres
import ast

def parse_genres(value):
    if pd.isna(value):
        return []
    try:
        parsed = ast.literal_eval(value)
        if isinstance(parsed, list):
            return [genre.strip().lower() for genre in parsed]
        return []
    except (ValueError, SyntaxError):
        return []

genres = df["genres"].apply(parse_genres).explode()
genres = genres.dropna()

print(genres.value_counts().head(10))
```

```python
# Keep titles with a recorded IMDb score
genre_scores = df.dropna(subset=["imdb_score"]).copy()

# Convert the genre strings into lists
genre_scores["genre_list"] = genre_scores["genres"].apply(parse_genres)

# Create one row per title-genre combination
genre_scores = genre_scores.explode("genre_list")

# Calculate average score and number of rated titles per genre
genre_summary = (
    genre_scores.dropna(subset=["genre_list"])
    .groupby("genre_list")["imdb_score"]
    .agg(["mean", "count"])
    .sort_values("mean", ascending=False)
    .round(2)
)

print(genre_summary.head(10))
```

table shows that History has the highest average IMDb score among the genres displayed, followed by War and Documentation.

```python
top_genres = genre_summary.head(10).sort_values("mean")

top_genres["mean"].plot(
    kind="barh",
    figsize=(9, 6)
)

plt.title("Top 10 Genres by Average IMDb Score")
plt.xlabel("Average IMDb Score")
plt.ylabel("Genre")
plt.xlim(0, 10)

plt.tight_layout()
plt.show()
```

```python
#rating analysis
df_rated = df.dropna(subset=["imdb_score"]).copy()

print("Original dataset:", len(df))
print("Dataset with IMDb scores:", len(df_rated))
print("Missing IMDb scores:", df_rated["imdb_score"].isnull().sum())
```

```python
#Compare average ratings by release decade
df_rated = df_rated.copy()

df_rated["decade"] = (
    (df_rated["release_year"] // 10) * 10
).astype(int)

decade_summary = (
    df_rated.groupby("decade")["imdb_score"]
    .agg(["mean", "count"])
    .round(2)
)

print(decade_summary)
```

```python
#bar chart to make the differences easier to compare
decade_summary["mean"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Average IMDb Score by Release Decade")
plt.xlabel("Release Decade")
plt.ylabel("Average IMDb Score")
plt.ylim(0, 10)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#top 10 highest-rated titles,
top_titles = (
    df.dropna(subset=["imdb_score", "imdb_votes"])
    .query("imdb_votes >= 1000")
    .sort_values("imdb_score", ascending=False)
    [["title", "type", "release_year", "imdb_score", "imdb_votes"]]
    .head(10)
)

print(top_titles.to_string(index=False))
```

```python
#verify whether movies appear among the top-rated titles at all when we apply the same vote threshold.
top_movies = (
    df.dropna(subset=["imdb_score", "imdb_votes"])
    .query("type == 'MOVIE' and imdb_votes >= 1000")
    .sort_values("imdb_score", ascending=False)
    [["title", "release_year", "imdb_score", "imdb_votes"]]
    .head(10)
)

print(top_movies.to_string(index=False))
```
